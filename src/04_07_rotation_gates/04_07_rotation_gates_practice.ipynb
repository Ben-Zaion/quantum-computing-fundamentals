Parameterized rotation gates (Rx, Ry, Rz) allow you to rotate a qubit by any angle around the x, y, or z-axis on the Bloch sphere, giving precise control over qubit states.<br><br>
When these gates rotate by pi radians (180 degrees), their effect is similar to the Pauli X, Y, and Z gates but differs by a global phase factor, which doesn't affect the measurable state.<br><br>
These gates are represented by specific matrices involving sine, cosine, and complex exponential functions, reflecting their continuous and adjustable nature.

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector
from math import pi
%matplotlib inline
from qiskit.quantum_info import Statevector
```

```python
circuit = QuantumCircuit(3)
circuit.rx(pi/3, 0)
circuit.ry(pi/3, 1)
circuit.h(2)
circuit.rz(pi/3, 2)
circuit.draw(output='mpl');
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```

```python
circuit = QuantumCircuit(3)
circuit.rx(2, 0)
circuit.ry(-pi/3, 1)
circuit.h(2)
circuit.rz(pi/3, 2)
circuit.draw(output='mpl');
```

```python
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```
