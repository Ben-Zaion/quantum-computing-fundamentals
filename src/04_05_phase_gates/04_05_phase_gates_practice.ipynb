Phase shift gates like the S and T gates rotate qubits around the Bloch sphere's z-axis by specific angles: the S gate by 90 degrees (pi/2 radians) and the T gate by 45 degrees (pi/4 radians). <br><br>
The S-dagger and T-dagger gates perform the opposite rotations, effectively undoing the phase shifts of the S and T gates. <br><br>
These gates are represented by specific matrices involving complex numbers, and their combinations follow mathematical rules, such as applying the S gate twice equating to the Pauli Z gate.

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector
%matplotlib inline
from qiskit.quantum_info import Statevector
```

```python
circuit = QuantumCircuit(3)
circuit.h([0, 1, 2])
circuit.s(1)
circuit.sdg(2)
#circuit.t(1)
#circuit.tdg(2)
circuit.draw(output='mpl');
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```

```python
circuit = QuantumCircuit(3)
circuit.h([0, 1, 2])
circuit.t(1)
circuit.tdg(2)
circuit.draw(output='mpl');
```

```python
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```
