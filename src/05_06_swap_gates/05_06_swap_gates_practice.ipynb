# SWAP Gate
Swaps the quantum state of two qubits

### Fredkin (Controlled-SWAP) Gate
Conditionally swaps two target qubits if a control qubit is | 1>

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector
from math import pi
%matplotlib inline
from qiskit.quantum_info import Statevector
```

```python
circuit = QuantumCircuit(3)
circuit.h(1)
circuit.rx(-pi/4, 2)
circuit.swap(1, 2)
circuit.draw(output='mpl');
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```

```python
circuit = QuantumCircuit(3)
circuit.h(1)
circuit.rx(-pi/4, 2)
circuit.cswap(0, 1, 2)
circuit.draw(output='mpl');
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```
