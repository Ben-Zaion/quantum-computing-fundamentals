# Toffoli (CCNOT) Gate
Conditionally flips the target qubit if *both* control qubits are 1> <br><br>
Abbreviated as CCNOT or CCX

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector
from math import pi
%matplotlib inline
from qiskit.quantum_info import Statevector
```

```python
circuit = QuantumCircuit(3)
circuit.x(0)
circuit.ccx(0, 1, 2)
circuit.draw(output='mpl');
#target not filp 
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```

```python
circuit = QuantumCircuit(3)
circuit.x(0)
circuit.x(1)
circuit.ccx(0, 1, 2)
circuit.draw(output='mpl');
#target filp. both control qubits = 1
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```

```python
circuit = QuantumCircuit(3)
circuit.x(0)
circuit.x(1)
circuit.rx(pi/3, 2)
circuit.ccx(0, 1, 2)
circuit.draw(output='mpl');
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
plot_bloch_multivector(statevector);
```
