# Pauli Gates


X = |0 1| bitlip <br>
&emsp;&emsp;|1 0| <br><br>
Y = |0 -i| a bit flip and phase flip <br>
&emsp;&emsp;|i 0| <br><br>
Z = |1 0| phase flip <br>
&emsp;&emsp;|0 -1|

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import array_to_latex,plot_bloch_multivector
from qiskit import *
from qiskit.visualization import plot_state_qsphere
%matplotlib inline

# Test X gate
circuit_x = QuantumCircuit(2)
circuit_x.x(0)
state_x = Statevector.from_instruction(circuit_x)
print("X Gate:")
display(array_to_latex(state_x))
circuit_x.draw(output='mpl');

#circuit_x.remove_final_measurements()  # no measurements allowed
#statevector_x = Statevector(circuit_x)
#plot_bloch_multivector(statevector_x)
```

```python
# Test Y gate
circuit_y = QuantumCircuit(1)
circuit_y.y(0)
state_y = Statevector.from_instruction(circuit_y)
print("Y Gate:")
display(array_to_latex(state_y))
circuit_y.draw(output='mpl');
```

```python
# Test Z gate
circuit_z = QuantumCircuit(1)
circuit_z.z(0)
state_z = Statevector.from_instruction(circuit_z)
print("Z Gate:")
display(array_to_latex(state_z))
circuit_z.draw(output='mpl');
```
