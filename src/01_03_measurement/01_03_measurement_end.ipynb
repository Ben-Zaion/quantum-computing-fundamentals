```python
from qiskit import *
from qiskit.visualization import plot_histogram

%matplotlib inline
```

```python
qr = QuantumRegister(1)
cr = ClassicalRegister(1)
circuit = QuantumCircuit(qr, cr)
# initialize a fresh qubit
circuit.h(0)
circuit.measure(qr, cr)
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
from qiskit.providers.basic_provider import BasicProvider
# Qiskit's qasm simulator backend which mimics the behavior of an actual quantum device 
# and gives back a dictionary of resulting counts
simulator = BasicProvider().get_backend('basic_simulator')
compiled_circuit = transpile(circuit, simulator)
result = simulator.run(compiled_circuit, shots=102400).result()
plot_histogram(result.get_counts())
```

```python
print(result.get_counts())
```

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

# Create a statevector simulator
simulator = AerSimulator(method='statevector')

# Save the final statevector of the circuit
circuit.save_statevector()

# Compile the circuit so the simulator can run it
compiled_circuit = transpile(circuit, simulator)

# Run the circuit and get the results
result = simulator.run(compiled_circuit).result()

# Retrieve the statevector from the results
statevector = result.get_statevector()

# Display the statevector
print(statevector)
```
