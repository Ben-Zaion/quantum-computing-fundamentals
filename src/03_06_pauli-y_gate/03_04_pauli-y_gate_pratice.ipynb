# Pauli-Y gate


Y|0> =  |0 -i| |1| --> |0x1 + -ix0| --> |0| = --> i|0| = i|1> <br>
&emsp;&emsp;&emsp;        |i   0| |0|&emsp;&emsp;&ensp;     |ix1 +  0x0|&emsp;&emsp;     |i|&emsp;&emsp;&emsp;&ensp;        |1| <br>
P(Y|0> = |0>) = |0|^2 = 0 <br>
P(Y|1> = |1>) = |i|^2 = 1 <br>
|1> is physically indistinguishable from i|1›

Y|1> =  |0 -i| |0| --> |0x0 + -ix1| --> |-i| = --> -i|1| = -i|0> <br>
&emsp;&emsp;&emsp;        |i   0| |1|&emsp;&emsp;&ensp;     |ix0 +  0x1|&emsp;&emsp;     |0|&emsp;&emsp;&emsp;&ensp;        |0| <br>
P(Y|1> = |0>) = |-i|^2 = 1 <br>
P(Y|1> = |1>) = |0|^2 = 0 <br>

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector, array_to_latex
from math import pi
%matplotlib inline
from qiskit.quantum_info import Statevector
```

```python
circuit = QuantumCircuit(2)
circuit.ry(pi/4, [0, 1])
```

```python
circuit.draw(output='mpl')
```

```python
# UPDATE: Code updated after recording to resolve breaking changes from Qiskit 1.0
circuit.remove_final_measurements()  # no measurements allowed
statevector = Statevector(circuit)
array_to_latex(statevector, prefix="\\text{statevector = }\n")
```

```python
plot_bloch_multivector(statevector)
```

```python
# apply Y gate
circuit.y(1)
```
